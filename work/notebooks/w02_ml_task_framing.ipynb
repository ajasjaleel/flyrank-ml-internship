## 1. My lane as an ML task (type)

### Task type: Ranking

I frame my lane as a **ranking** problem. The goal is to rank content opportunities by their expected value so that the highest-priority opportunities can be reviewed and acted on first.

Ranking fits this problem because the practical decision is not only whether an opportunity is good or bad, but **which opportunities should be prioritized first**. The model can use multiple signals from the available data to produce an ordered list for decision support.


```python
task_type = "ranking"
print("ML task type:", task_type)
```

## 2. Target or proxy

### Target or proxy

I would predict a **priority score for each content opportunity**, representing how strongly the opportunity should be considered for action. This would be a **proxy**, because the dataset does not directly observe the future business value of a content refresh.

The proxy would be defined from measurable signals in the available data rather than from a known future outcome. I would use it for **decision support**, not as a claim that the model can perfectly predict future performance.


```python
target_type = "proxy"
target_name = "content opportunity priority score"

print("Target type:", target_type)
print("Target/proxy:", target_name)
```

## 3. Success metric

### Success metric

I would use **Precision@50** as the success metric. This measures the proportion of the top 50 ranked opportunities that are actually useful according to the defined outcome or proxy.

A higher Precision@50 means that more of the opportunities placed at the top of the ranking are relevant for action. This is useful because content teams have limited time and need to focus on the highest-priority opportunities first.


```python
metric_name = "Precision@50"

print("Success metric:", metric_name)
print("Good means: a higher proportion of the top 50 ranked opportunities are useful.")
```

## 5. Why ML beats a fixed rule here

### Why ML beats a fixed rule

A fixed rule would rely on one or two manually chosen thresholds, such as prioritizing an opportunity whenever search volume is above a certain value. This can miss useful opportunities because content performance depends on multiple signals and their combinations.

ML is useful here because it can learn patterns across several measurable features instead of relying on one fixed threshold. The resulting ranking can provide **decision support** by helping prioritize opportunities, while still requiring human review before action.


```python
reason = "multiple signals and their combinations are harder to capture with one fixed threshold"

print("Why ML:", reason)
```
